```python
import numpy as np
import matplotlib.pyplot as plt
import time

np.random.seed(42)

print("NumPy Version:", np.__version__)
```

```python
import matplotlib.cbook as cbook

with cbook.get_sample_data("grace_hopper.jpg") as image_file:
    image = plt.imread(image_file)

print("Image Loaded Successfully")
print("Image Shape:", image.shape)
print("Image Data Type:", image.dtype)

plt.figure(figsize=(8, 6))
plt.imshow(image)
plt.title("Original Image")
plt.axis("off")
plt.show()
```

```python
image_float = image[..., :3].astype(np.float32)

grayscale = (
    0.299 * image_float[:, :, 0] +
    0.587 * image_float[:, :, 1] +
    0.114 * image_float[:, :, 2]
)

print("Original Shape:", image.shape)
print("Grayscale Shape:", grayscale.shape)

plt.figure(figsize=(8, 6))
plt.imshow(grayscale, cmap="gray")
plt.title("Grayscale Image")
plt.axis("off")
plt.show()
```

```python
horizontal_flip = np.flip(image, axis=1)

vertical_flip = np.flip(image, axis=0)

plt.figure(figsize=(12, 5))

plt.subplot(1, 2, 1)
plt.imshow(horizontal_flip)
plt.title("Horizontal Flip")
plt.axis("off")

plt.subplot(1, 2, 2)
plt.imshow(vertical_flip)
plt.title("Vertical Flip")
plt.axis("off")

plt.show()
```

```python
rotated_image = np.rot90(image)

print("Original Shape:", image.shape)
print("Rotated Shape:", rotated_image.shape)

plt.figure(figsize=(7, 6))
plt.imshow(rotated_image)
plt.title("Rotated Image - 90 Degrees")
plt.axis("off")
plt.show()
```
